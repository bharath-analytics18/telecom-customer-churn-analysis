```python
import pandas as pd

df = pd.read_csv('telco_churn_raw.csv')
df.head()
```

```python
df.info()
```

```python
df[df['TotalCharges'] == ' ']
```

```python
df['TotalCharges'] = df['TotalCharges'].replace(' ', '0')
df['TotalCharges'] = df['TotalCharges'].astype(float)
df.info()
```

```python
def tenure_group(t):
    if t <= 12:
        return '0-12 months'
    elif t <= 24:
        return '12-24 months'
    elif t <= 48:
        return '24-48 months'
    elif t <= 60:
        return '48-60 months'
    else:
        return '60+ months'

df['TenureGroup'] = df['tenure'].apply(tenure_group)
df[['tenure', 'TenureGroup']].head(10)
```

```python
df.to_csv('telco_churn_cleaned.csv', index=False)
print("Cleaned file saved!")
```

```python
churn_counts = df['Churn'].value_counts()
churn_rate = df['Churn'].value_counts(normalize=True) * 100

print(churn_counts)
print()
print(churn_rate.round(1))
```

```python
contract_churn = df.groupby('Contract')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).round(1)
print(contract_churn)
```

```python
tenure_churn = df.groupby('TenureGroup')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).round(1)
print(tenure_churn)
```

```python
payment_churn = df.groupby('PaymentMethod')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).round(1)
print(payment_churn)
```

```python
techsupport_churn = df.groupby('TechSupport')['Churn'].apply(lambda x: (x == 'Yes').mean() * 100).round(1)
print(techsupport_churn)
```

```python
avg_charges = df.groupby('Churn')['MonthlyCharges'].mean().round(2)
print(avg_charges)
```
